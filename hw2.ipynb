```python
import numpy as np


def softmax(x):
    if x.ndim == 2:
        c = np.max(x, axis=1, keepdims=True)
        exp_x = np.exp(x - c)
        return exp_x / np.sum(exp_x, axis=1, keepdims=True)

    c = np.max(x)
    exp_x = np.exp(x - c)
    return exp_x / np.sum(exp_x)

def cross_entropy_error(y_hat, y):

    if y_hat.ndim == 1:
        y_hat = y_hat.reshape(1, y_hat.size)
        y = y.reshape(1, y.size)

    batch_size = y_hat.shape[0]

    delta = 1e-7

    return -np.sum(y * np.log(y_hat + delta)) / batch_size
```

```python

X = np.array([
    [0.1, 0.5, -0.2, 0.8, 0.0, -1.1],
    [0.3, -0.2, 0.4, 0.1, 0.5, 0.2],
    [-0.5, 0.2, 0.9, -0.3, 0.4, 0.1],
    [0.2, 0.4, 0.1, -0.5, -0.2, 0.3],
    [-0.1, 0.0, 0.3, 0.6, -0.4, -0.2]
])

Y = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [1, 0, 0],
    [0, 1, 0]
])

print(f"X 的形狀 (Shape): {X.shape}")
print(f"Y 的形狀 (Shape): {Y.shape}")
```

```python


np.random.seed(42)


W1 = np.random.randn(6, 5)
b1 = np.random.randn(5)


W2 = np.random.randn(5, 4)
b2 = np.random.randn(4)


W3 = np.random.randn(4, 3)
b3 = np.random.randn(3)

print("--- 各層權重與偏值形狀 ---")
print(f"W1 shape: {W1.shape}, b1 shape: {b1.shape}")
print(f"W2 shape: {W2.shape}, b2 shape: {b2.shape}")
print(f"W3 shape: {W3.shape}, b3 shape: {b3.shape}")
```

```python

z1 = np.dot(X, W1) + b1
h1 = np.maximum(0, z1)


z2 = np.dot(h1, W2) + b2
h2 = np.maximum(0, z2)


z3 = np.dot(h2, W3) + b3
y_hat = softmax(z3)

print("--- 各層加權和 z 與輸出向量形狀 ---")
print(f"z1 (隱藏層1加權和) shape: {z1.shape}")
print(f"h1 (隱藏層1輸出) shape: {h1.shape}")
print(f"z2 (隱藏層2加權和) shape: {z2.shape}")
print(f"h2 (隱藏層2輸出) shape: {h2.shape}")
print(f"z3 (輸出層加權和) shape: {z3.shape}")
print(f"y_hat (最終預設機率輸出) shape: {y_hat.shape}")
print("\ny_hat 結果預覽:")
print(y_hat)
```

```python

loss = cross_entropy_error(y_hat, Y)
print(f"整批 (5筆資料) 的平均 Cross Entropy Error: {loss:.4f}")
```
