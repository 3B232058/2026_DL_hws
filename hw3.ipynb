```python
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from sklearn.metrics import confusion_matrix
import seaborn as sns
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Dropout, Flatten
from tensorflow.keras.utils import to_categorical

class_names = [
    'T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]


(x_train_raw, y_train_raw), (x_test_raw, y_test_raw) = fashion_mnist.load_data()


print("=== 資料形狀檢視 ===")
print(f"訓練資料 x_train 形狀: {x_train_raw.shape}")
print(f"訓練標籤 y_train 形狀: {y_train_raw.shape}")
print(f"測試資料 x_test  形狀: {x_test_raw.shape}")
print(f"測試標籤 y_test  形狀: {y_test_raw.shape}")


plt.figure(figsize=(10, 4))
for i in range(3):
    plt.subplot(1, 3, i + 1)
    plt.imshow(x_train_raw[i], cmap='gray')
    label_index = y_train_raw[i]
    plt.title(f"Label: {class_names[label_index]}\n(Class {label_index})")
    plt.axis('off')
plt.suptitle("訓練資料範例 (前三張)", fontsize=14)
plt.tight_layout()
plt.show()

x_train = x_train_raw.astype('float32') / 255.0
x_test = x_test_raw.astype('float32') / 255.0


y_train = to_categorical(y_train_raw, 10)
y_test = to_categorical(y_test_raw, 10)




model = Sequential([
    Flatten(input_shape=(28, 28), name='Flatten_Input'),
    Dense(512, activation='relu', name='Dense_Hidden_1'),
    Dropout(0.2, name='Dropout_1'),
    Dense(256, activation='relu', name='Dense_Hidden_2'),
    Dropout(0.2, name='Dropout_2'),
    Dense(10, activation='softmax', name='Dense_Output')
])

print("\n=== 模型架構摘要 ===")
model.summary()

"""
【每一層參數個數計算說明】：
1. Flatten_Input 層:
   - 將 28x28 圖片展平為 784 維向量。沒有可訓練參數 (Param = 0)。

2. Dense_Hidden_1 層 (512 神經元):
   - 權重 (Weights): 784 (輸入) × 512 (神經元) = 401,408
   - 偏置 (Biases): 512
   - 總參數 = 401,408 + 512 = 401,920

3. Dropout_1 層:
   - 隨機捨棄神經元以防止 Overfitting。無可訓練參數 (Param = 0)。

4. Dense_Hidden_2 層 (256 神經元):
   - 權重 (Weights): 512 (輸入) × 256 (神經元) = 131,072
   - 偏置 (Biases): 256
   - 總參數 = 131,072 + 256 = 131,328

5. Dropout_2 層:
   - 無可訓練參數 (Param = 0)。

6. Dense_Output 層 (10 神經元):
   - 權重 (Weights): 256 (輸入) × 10 (神經元) = 2,560
   - 偏置 (Biases): 10
   - 總參數 = 2,560 + 10 = 2,570

全模型總參數 (Total Params) = 401,920 + 131,328 + 2,570 = 535,818
"""



model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)


batch_size = 128
epochs = 20

print("\n=== 批次輸入與各層輸出資料形狀 ===")
print(f"• 批次輸入 (Input Shape)  : (batch_size, 28, 28) -> ({batch_size}, 28, 28)")
print(f"• Flatten 輸出形狀         : ({batch_size}, 784)")
print(f"• Dense_Hidden_1 輸出形狀 : ({batch_size}, 512)")
print(f"• Dense_Hidden_2 輸出形狀 : ({batch_size}, 256)")
print(f"• Dense_Output 輸出形狀   : ({batch_size}, 10)")


history = model.fit(
    x_train, y_train,
    batch_size=batch_size,
    epochs=epochs,
    validation_split=0.2,
    verbose=1
)




plt.figure(figsize=(14, 5))


plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy', linewidth=2)
plt.plot(history.history['val_accuracy'], label='Val Accuracy', linewidth=2)
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)


plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss', linewidth=2)
plt.plot(history.history['val_loss'], label='Val Loss', linewidth=2)
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

print("\n【Overfitting 評估說明】：")
print("觀察 Training 與 Validation 的曲線。加入 Dropout 層後，Validation Loss 與 Training Loss 保持貼近趨勢，代表過度擬合 (Overfitting) 已獲得有效控制。")




test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)
print(f"\n=== 測試集評估結果 ===")
print(f"測試集 Loss: {test_loss:.4f}")
print(f"測試集 Accuracy: {test_acc * 100:.2f}%")

if test_acc >= 0.90:
    print("✅ 成功達成目標：準確率達到 90% 以上！")
else:
    print("⚠️ 準確率未達 90%，建議調整 Epoch 次數或增加神經元數量。")





y_pred_probs = model.predict(x_test)
y_pred = np.argmax(y_pred_probs, axis=1)


plt.figure(figsize=(12, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test_raw[i], cmap='gray')
    true_label = class_names[y_test_raw[i]]
    pred_label = class_names[y_pred[i]]

    color = 'green' if y_test_raw[i] == y_pred[i] else 'red'
    plt.title(f"True: {true_label}\nPred: {pred_label}", color=color)
    plt.axis('off')

plt.suptitle("測試資料預測結果展示", fontsize=14)
plt.tight_layout()
plt.show()



cm = confusion_matrix(y_test_raw, y_pred)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Fashion MNIST Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()




model.save('fashion_mnist_dnn.h5')
print("\n=== 模型儲存成功 ===")
print("完整模型已成功儲存為 'fashion_mnist_dnn.h5'。")
```
