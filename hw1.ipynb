```python
import numpy as np
import gradio as gr

np.random.seed(42)

x = np.array([1.2, -0.5, 2.3, 0.8, -1.1, 0.4])
print("輸入資料 x:")
print(f"形狀 (Shape): {x.shape}")
print(f"數值: {x}\n")
```

```python
def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def softmax(z):
    exp_z = np.exp(z - np.max(z))
    return exp_z / np.sum(exp_z)
```

```python

W1 = np.array([
    [0.1, -0.2, 0.3, 0.1, 0.2],
    [-0.4, 0.5, -0.1, 0.3, 0.0],
    [0.2, 0.1, -0.3, 0.4, -0.2],
    [-0.3, -0.1, 0.2, -0.1, 0.5],
    [0.5, 0.3, 0.1, -0.2, -0.4],
    [-0.1, 0.2, 0.4, 0.1, 0.3]
])
b1 = np.array([0.1, -0.1, 0.2, 0.0, -0.2])

z1 = np.dot(x, W1) + b1
h1 = relu(z1)

print("【第一隱藏層】")
print(f"W1 形狀: {W1.shape}，參數個數: {W1.size}")
print(f"b1 形狀: {b1.shape}，參數個數: {b1.size}")
print(f"調整後加權和 z1: {z1}，形狀: {z1.shape}")
print(f"輸出向量 h1: {h1}，形狀: {h1.shape}\n")


W2 = np.array([
    [0.2, -0.3, 0.1, 0.4],
    [-0.1, 0.4, 0.2, -0.2],
    [0.3, -0.2, -0.4, 0.1],
    [0.1, 0.1, 0.3, -0.3],
    [-0.2, 0.3, 0.0, 0.2]
])
b2 = np.array([0.0, 0.1, -0.1, 0.2])

z2 = np.dot(h1, W2) + b2
h2 = relu(z2)

print("【第二隱藏層】")
print(f"W2 形狀: {W2.shape}，參數個數: {W2.size}")
print(f"b2 形狀: {b2.shape}，參數個數: {b2.size}")
print(f"調整後加權和 z2: {z2}，形狀: {z2.shape}")
print(f"輸出向量 h2: {h2}，形狀: {h2.shape}\n")


W3 = np.array([
    [0.4, -0.1, 0.2, 0.1],
    [-0.3, 0.2, -0.2, 0.3],
    [0.1, 0.3, 0.4, -0.1],
    [-0.2, -0.4, 0.1, 0.2]
])
b3 = np.array([-0.1, 0.0, 0.2, 0.1])

z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)

print("【第三隱藏層】")
print(f"W3 形狀: {W3.shape}，參數個數: {W3.size}")
print(f"b3 形狀: {b3.shape}，參數個數: {b3.size}")
print(f"調整後加權和 z3: {z3}，形狀: {z3.shape}")
print(f"輸出向量 h3: {h3}，形狀: {h3.shape}\n")


W4 = np.array([
    [0.2, -0.4, 0.1],
    [-0.1, 0.3, 0.2],
    [0.4, 0.1, -0.3],
    [-0.3, -0.2, 0.5]
])
b4 = np.array([0.1, -0.2, 0.0])

z4 = np.dot(h3, W4) + b4
y_hat = softmax(z4)

print("【輸出層】")
print(f"W4 形狀: {W4.shape}，參數個數: {W4.size}")
print(f"b4 形狀: {b4.shape}，參數個數: {b4.size}")
print(f"調整後加權和 z4 (logits): {z4}，形狀: {z4.shape}")
print(f"輸出向量 y_hat: {y_hat}，形狀: {y_hat.shape}\n")
```

```python
y = np.array([0, 0, 1])

delta = 1e-7
cross_entropy_error = -np.sum(y * np.log(y_hat + delta))

print("【誤差計算】")
print(f"真實結果 y: {y}")
print(f"預測結果 y_hat: {y_hat}")
print(f"Cross Entropy Error: {cross_entropy_error:.6f}")
```

```python
def dnn(x_input):
    """接收 6 個特徵值的陣列，執行 DNN 前向傳播並回傳 y_hat"""
    x_arr = np.array(x_input, dtype=float)

    z1 = np.dot(x_arr, W1) + b1
    h1 = relu(z1)

    z2 = np.dot(h1, W2) + b2
    h2 = relu(z2)

    z3 = np.dot(h2, W3) + b3
    h3 = sigmoid(z3)

    z4 = np.dot(h3, W4) + b4
    y_hat_result = softmax(z4)

    return y_hat_result

result_test = dnn(x)
print("使用 dnn() 函數計算的 y_hat:")
print(result_test)
print(f"與原本結果是否完全相同: {np.allclose(y_hat, result_test)}")
```

```python
def predict_class(f1, f2, f3, f4, f5, f6):

    x_input = [f1, f2, f3, f4, f5, f6]


    probabilities = dnn(x_input)


    max_index = np.argmax(probabilities)
    predicted_category = max_index + 1


    output_str = f"各類別機率 (y_hat):\n" \
                 f"  - 類別 1: {probabilities[0]:.4f}\n" \
                 f"  - 類別 2: {probabilities[1]:.4f}\n" \
                 f"  - 類別 3: {probabilities[2]:.4f}\n\n" \
                 f"👉 機率最高的類別為：【 類別 {predicted_category} 】"

    return output_str

demo = gr.Interface(
    fn=predict_class,
    inputs=[
        gr.Number(value=1.2, label="特徵值 1"),
        gr.Number(value=-0.5, label="特徵值 2"),
        gr.Number(value=2.3, label="特徵值 3"),
        gr.Number(value=0.8, label="特徵值 4"),
        gr.Number(value=-1.1, label="特徵值 5"),
        gr.Number(value=0.4, label="特徵值 6")
    ],
    outputs=gr.Textbox(label="預測結果"),
    title="DNN 多隱藏層神經網路分類器",
    description="請輸入 6 個特徵值，系統將透過多層感知機計算預測類別與對應機率。"
)


demo.launch(inline=False)
```
